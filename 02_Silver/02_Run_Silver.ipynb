```python
# ============================================================
# 02_Run_Silver
# Parent / Orchestrator Notebook
# ============================================================

datasets = {

    "drivers.csv": "drivers",

    "constructors.csv": "constructors",

    "circuits.csv": "circuits",

    "races.csv": "races",

    "results.csv": "results",

    "qualifying.csv": "qualifying",

    "pit_stops.csv": "pit_stops",

    "lap_times.csv": "lap_times",

    "driver_standings.csv": "driver_standings",

    "constructor_standings.csv": "constructor_standings",

    "seasons.csv": "seasons",

    "status.csv": "status"
}


# ============================================================
# RESULT COLLECTION
# ============================================================

silver_results = []


# ============================================================
# LOOP THROUGH ALL DATASETS
# ============================================================

for file_name, table_name in datasets.items():

    print("\n" + "=" * 80)

    print(
        f"Starting Silver transformation: "
        f"{table_name}"
    )

    print("=" * 80)


    try:

        result = dbutils.notebook.run(

            "02_Silver_Transformation",

            0,

            {
                "table_name": table_name
            }
        )


        silver_results.append({

            "file_name": file_name,

            "table_name": table_name,

            "status": "SUCCESS",

            "message": result
        })


        print(
            f"SUCCESS: {table_name}"
        )


    except Exception as e:

        silver_results.append({

            "file_name": file_name,

            "table_name": table_name,

            "status": "FAILED",

            "message": str(e)
        })


        print(
            f"FAILED: {table_name}"
        )

        print(
            f"Error: {str(e)}"
        )


# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n")
print("=" * 80)
print("SILVER PROCESSING SUMMARY")
print("=" * 80)


for result in silver_results:

    print(
        f"{result['table_name']:25} "
        f"{result['status']:10} "
        f"{result['message']}"
    )


print("=" * 80)


# ============================================================
# CHECK IF ANY TABLE FAILED
# ============================================================

failed_tables = [

    result
    for result in silver_results
    if result["status"] == "FAILED"
]


if failed_tables:

    print(
        f"\nSilver processing completed with "
        f"{len(failed_tables)} failure(s)."
    )

else:

    print(
        "\nAll Silver tables processed successfully."
    )
```
