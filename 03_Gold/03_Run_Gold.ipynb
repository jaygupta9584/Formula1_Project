```python
# ============================================================
# 03_Run_Gold
# Gold Parent / Orchestrator
# ============================================================


gold_datasets = [

    "driver_performance",

    "constructor_performance",

    "season_summary",

    "circuit_performance"
]


# ============================================================
# RESULT COLLECTION
# ============================================================

gold_results = []


# ============================================================
# RUN GOLD DATASETS
# ============================================================

for dataset_name in gold_datasets:

    print("\n" + "=" * 80)

    print(
        f"Starting Gold transformation: "
        f"{dataset_name}"
    )

    print("=" * 80)


    try:

        result = dbutils.notebook.run(

            "03_Gold_Transformation",

            0,

            {
                "dataset_name": dataset_name
            }
        )


        gold_results.append({

            "dataset_name": dataset_name,

            "status": "SUCCESS",

            "message": result

        })


        print(
            f"SUCCESS: {dataset_name}"
        )


    except Exception as e:

        gold_results.append({

            "dataset_name": dataset_name,

            "status": "FAILED",

            "message": str(e)

        })


        print(
            f"FAILED: {dataset_name}"
        )

        print(
            f"Error: {str(e)}"
        )


# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n")

print("=" * 80)

print("GOLD PROCESSING SUMMARY")

print("=" * 80)


for result in gold_results:

    print(
        f"{result['dataset_name']:30} "
        f"{result['status']:10} "
        f"{result['message']}"
    )


print("=" * 80)


# ============================================================
# FAILURE CHECK
# ============================================================

failed = [

    result
    for result in gold_results
    if result["status"] == "FAILED"
]


if failed:

    print(
        f"Gold processing completed with "
        f"{len(failed)} failure(s)."
    )

else:

    print(
        "All Gold datasets processed successfully."
    )
```
