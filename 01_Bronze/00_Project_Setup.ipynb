```python
# Create Catalog
spark.sql("""
CREATE CATALOG IF NOT EXISTS f1_catalog
""")

# Create Bronze
spark.sql("""
CREATE SCHEMA IF NOT EXISTS f1_catalog.bronze
""")

# Create Silver
spark.sql("""
CREATE SCHEMA IF NOT EXISTS f1_catalog.silver
""")

# Create Gold
spark.sql("""
CREATE SCHEMA IF NOT EXISTS f1_catalog.gold
""")

# Create raw volume
spark.sql("""
CREATE VOLUME IF NOT EXISTS f1_catalog.bronze.raw
""")

print("F1 project structure created successfully!")
```
